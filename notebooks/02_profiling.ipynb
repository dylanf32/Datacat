### Missing values

```python
def missing_val(df):

    print("Press 0 to remove missing values or Press 1 to fill them(will be filled with mean)")
    choice = int(input())

    if choice == 0:

        df.isna()

        df = df.dropna()

    elif choice == 1:

        df= df.fillna().mean()

    return df
```

#### Duplicates

```python
def duplicates(df):

    print("would you like to remove duplicates: 1- yes 2-no")
    choice_2 = int(input())

    if choice_2 == 1:

        df= df.drop_duplicates()

    elif choice_2 == 2: 
        print("ok!")
    else:
        print("wrong choice buddy")

    return df

```

### Descriptive stats

```python
def describe_data(df):

    print(df.describe())    
    print(df.sum())    
    print(df.cumprod())    
    print(df.count())    

    return df
```
